- An agent is an object that can move in the maze to find a path to the goal cell.

3. Pathfinding algorithms:
- A-star: an informed search algorithm that takes both the cost and a heuristic into account. It uses the score f(n) = g(n) + h(n), where g(n) is the cost from the start node to node n, and h(n) is an estimate of the distance from node n to the goal. A* picks the nodes to expand by their f(n) value, preferring nodes with a low f(n).
- BFS (Breadth-First Search): BFS searches by breadth. It explores all the nodes at the same level before going deeper. In a maze, BFS can be used to find the shortest path from the start node to the goal node.
- DFS (Depth-First Search): DFS searches by depth. It goes as deep as it can down one branch of the graph before backing up and trying another branch. In a maze, DFS can be used to check whether a path from the start node to the goal node exists.
- Dijkstra: an algorithm that finds the shortest path. It visits the nodes by breadth and updates the cost from the start node to the neighbouring nodes. In a maze, Dijkstra can find the shortest path, but it can become slow when the edges have different costs

4. Path and path tracing:

- When the agent moves, it leaves footprints on the cell it came from.
- It can trace back from the goal cell to the start cell and draw the shortest path.

#### a. Agent

```python
class agent: #The agents can be placed on the maze to indcate the cell selected in Maze
    def __init__(self,parentMaze,x=None,y=None,goal=None,filled=False,footprints=False,color:COLOR=COLOR.blue):
        # x,y-->  agent position, default is the last cell (rows, cols) of the maze
        # goal-->     default is (1,1)
        # footprints->When agent move to other cell, its footprints on the previous cell can be placed by making this True
        # color-->    color of the agent and of its footprints
        # position--> current coordinates of the agent (x,y)
        self._parentMaze=parentMaze #the maze the agent stands on
        self.color=color # pick the color for the agent
        self.filled=filled # agent size: true = big, false = small
        if x is None:x=parentMaze.rows # if no start position is given for the agent
        if y is None:y=parentMaze.cols # then use [rows, cols] --> the last cell of the maze
        self.x=x
        self.y=y
        self.footprints=footprints
        self._parentMaze._agents.append(self) # add the agent to the maze's agent list
        if goal==None: self.goal=self._parentMaze._goal # if goal = None, use the default goal (1,1)
        else:          self.goal=goal
        self._body=[] # keeps track of the agent's previous positions
        self.position=(self.x,self.y) # set the agent's start position

    @property # like a getter in Java
    def x(self): return self._x
    @x.setter # like a setter in Java
    def x(self,newX): self._x=newX
    @property
    def y(self): return self._y
    @y.setter
    def y(self,newY):
        self._y=newY
        w=self._parentMaze._cell_width # w is the size of one cell
        x=self.x*w-w+self._parentMaze._LabWidth
        y=self.y*w-w+self._parentMaze._LabWidth
        if self.filled:
                self._coord=(y, x,y + w, x + w) # draw the big square
        else:   # draw a small square, shrunk down from the big one
                self._coord=(y + w/2.5, x + w/2.5,y + w/2.5 +w/4, x + w/2.5 +w/4)
        #coord is the 4 numbers that fix one position on the tkinter canvas
        self._head=self._parentMaze._canvas.create_rectangle(*self._coord,fill=self.color.value[0],outline='')#stipple='gray75'
        # draw a square at coord --> this is the agent drawn on the maze by the canvas

        self._parentMaze._redrawCell(self.x,self.y,theme=self._parentMaze.theme)
    @property
    def position(self):
        return (self.x,self.y)
    @position.setter
    def position(self,newpos):
        self.x=newpos[0]
        self.y=newpos[1]
        self._position=newpos
```

#### c. Creating the maze

1. Setting up the variables
- _stack: keeps track of the cells while the maze is being built.
- _closed: keeps track of the cells already visited.
- self.theme: holds the colors of the maze.
- self._goal: holds the goal of the maze, chosen as (x, y) by default
2. Building the maze with a Recursive Backtracker
Setting up:
- cell = []: an empty list that holds the directions the current cell can go.
Loop over the directions (East, West, North, South) and check whether the neighbouring cell has not been visited yet and is inside the grid. If both hold, add that direction to the cell list.
- Pick a random direction:

   + If there is at least one direction in the cell list, pick one at random from the list.
   + Open the matching wall and update the path.
- Move and push the new cell onto the stack:

   + Move to the new cell in the chosen direction.
   + Add the new cell to the closed (visited) list.
   + Push the new cell onto the stack to carry on from there.
- Back up if there is no next direction:

   + If no direction is available, go back to the previous cell by popping it off the stack.
   + Repeat until the stack is empty, which means every reachable cell has been visited and the maze is finished, with random paths through it.

```python
class maze: #This is the main class to create maze.
    def __init__(self,rows=10,cols=10): #Need to pass just the two arguments. The rest will be assigned automatically
        self.rows=rows #rows--> No. of rows of the maze
        self.cols=cols #cols--> No. of columns of the maze
        self.maze_map={} #maze_map--> Will be set to a Dicationary
        # Keys will be cells and values will be another dictionary with keys=['E','W','N','S'] for
        # East West North South and values will be 0 or 1. 0 means direction(EWNS) is blocked, 1 means direction is open.
        self.grid=[] #grid--> A list of all cells
        self.path={} # Shortest path from start(bottom right) to goal(by default top left).It is a dictionary
        self._cell_width=50  #is cell width calculated automatically
        self._win=None # the window that will hold the maze
        self._canvas=None # the canvas object used to draw the maze
        self._agents=[] #_agents-->  A list of aganets on the maze
        self.markCells=[] #Will be used to mark some particular cell during path trace by the agent
    @property
    def grid(self): return self._grid
    @grid.setter
    def grid(self,n): # build the grid and give every cell four walls
        self._grid=[]
        y=0
        for n in range(self.cols): # cell coordinates start at (1,1) --> (row, col)
            x = 1
            y = 1+y
            for m in range(self.rows):
                self.grid.append((x,y))
                self.maze_map[x,y]={'E':0,'W':0,'N':0,'S':0}
                x = x + 1
    def _Open_East(self,x, y): # remove the wall on the East side
        self.maze_map[x,y]['E']=1 # open the E wall of this cell
        if y+1<=self.cols: # if there is still a cell on the right
            self.maze_map[x,y+1]['W']=1 # set that cell's W side to 1
    def _Open_West(self,x, y):
        self.maze_map[x,y]['W']=1
        if y-1>0:
            self.maze_map[x,y-1]['E']=1
    def _Open_North(self,x, y):
        self.maze_map[x,y]['N']=1
        if x-1>0:
            self.maze_map[x-1,y]['S']=1
    def _Open_South(self,x, y):
        self.maze_map[x,y]['S']=1
        if x+1<=self.rows:
            self.maze_map[x+1,y]['N']=1

    def CreateMaze(self,x=1,y=1,theme:COLOR=COLOR.dark):#important function to create a Random Maze
        _stack=[]
        _closed=[] # stores the visited cells
        self.theme=theme
        self._goal=(x,y) # default is (1,1)
        # generate a randomly maze
        _stack.append((x,y))
        _closed.append((x,y))
        while len(_stack) > 0:
            cell = []
            if(x , y +1) not in _closed and (x , y+1) in self.grid:     cell.append("E")
            if (x , y-1) not in _closed and (x , y-1) in self.grid:     cell.append("W")
            if (x+1, y ) not in _closed and (x+1 , y ) in self.grid:    cell.append("S")
            if (x-1, y ) not in _closed and (x-1 , y) in self.grid:     cell.append("N")
            if len(cell) > 0:
                current_cell = (random.choice(cell))
                if current_cell == "E":
                        self._Open_East(x,y)
                        self.path[x, y+1] = x, y
                        y = y + 1
                        _closed.append((x, y))
                        _stack.append((x, y))
                elif current_cell == "W":
                        self._Open_West(x, y)
                        self.path[x , y-1] = x, y
                        y = y - 1
                        _closed.append((x, y))
                        _stack.append((x, y))
                elif current_cell == "N":
                        self._Open_North(x, y)
                        self.path[(x-1 , y)] = x, y
                        x = x - 1
                        _closed.append((x, y))
                        _stack.append((x, y))
                elif current_cell == "S":
                        self._Open_South(x, y)
                        self.path[(x+1 , y)] = x, y
                        x = x + 1
                        _closed.append((x, y))
                        _stack.append((x, y))

            else:
                x, y = _stack.pop()
        ## Multiple Path Loops
        midx = self.rows // 2
        choices = [1] * 9 + [0] * 1
        for mid_cell_y in range(2,self.cols-2):
            t = random.choice(choices)
            if t :
               self._Open_North(midx, mid_cell_y)
               self._Open_South(midx, mid_cell_y)
               self._Open_West(midx-1, mid_cell_y)
               self._Open_East(midx+1, mid_cell_y)
        midy = self.cols // 2
        for mid_cell_x in range(2,self.rows-2):
            t = random.choice(choices)
            if t :
               self._Open_North(mid_cell_x, midy)
               self._Open_South(mid_cell_x, midy)
               self._Open_West(mid_cell_x, midy-1)
               self._Open_East(mid_cell_x, midy+1)
        self._drawMaze(self.theme)
        agent(self,*self._goal,filled=True,color=COLOR.green)

    def _drawMaze(self,theme): # draw the maze with tkinter
        self._LabWidth=26 # Space from the top for Labels, width of Label
        self._win=Tk()
        self._win.state('zoomed') # make the window fill the whole screen
        self._win.title('G6 INTRO AI') # window title
        scr_width=self._win.winfo_screenwidth() # default window width - depends on the machine (14 inch, 15 inch)
        scr_height=self._win.winfo_screenheight() # height of the window shown
        self._win.geometry(f"{scr_width}x{scr_height}+0+0") # make a window of size width x height
        self._canvas = Canvas(width=scr_width, height=scr_height, bg=theme.value[0]) # 0,0 is top left corner
        self._canvas.pack(expand=YES, fill=BOTH)
        # Some calculations for calculating the width of the maze cell
        k=3.25
        if self.rows>=95 and self.cols>=95:   k=0
        elif self.rows>=80 and self.cols>=80: k=1
        elif self.rows>=70 and self.cols>=70: k=1.5
        elif self.rows>=50 and self.cols>=50: k=2
        elif self.rows>=35 and self.cols>=35: k=2.5
        elif self.rows>=22 and self.cols>=22: k=3
        self._cell_width=round(min(((scr_height-self.rows-k*self._LabWidth)/(self.rows)),((scr_width-self.cols-k*self._LabWidth)/(self.cols)),90),3)
        # cell_width is capped at 90 and rounded to 3 decimals
        # Creating Maze lines
        if self._win is not None:
            if self.grid is not None:
                for cell in self.grid:
                    x,y=cell
                    w=self._cell_width #x,y,w --> used to compute the wall line coordinates for create_line
                    x=x*w-w+self._LabWidth
                    y=y*w-w+self._LabWidth
                    if self.maze_map[cell]['E']==False:
                        l=self._canvas.create_line(y + w, x, y + w, x + w,width=2,fill=theme.value[1],tag='line')
                    if self.maze_map[cell]['W']==False:
                        l=self._canvas.create_line(y, x, y, x + w,width=2,fill=theme.value[1],tag='line')
                    if self.maze_map[cell]['N']==False:
                        l=self._canvas.create_line(y, x, y + w, x,width=2,fill=theme.value[1],tag='line')
                    if self.maze_map[cell]['S']==False:
                        l=self._canvas.create_line(y, x + w, y + w, x + w,width=2,fill=theme.value[1],tag='line')

    def _redrawCell(self,x,y,theme):
        # To redraw a cell. With Full sized square agent, it can overlap with maze lines
        # So the cell is redrawn so that cell lines are on top
        w=self._cell_width
        cell=(x,y)
        x=x*w-w+self._LabWidth
        y=y*w-w+self._LabWidth
        if self.maze_map[cell]['E']==False:
            self._canvas.create_line(y + w, x, y + w, x + w,width=2,fill=theme.value[1])
        if self.maze_map[cell]['W']==False:
            self._canvas.create_line(y, x, y, x + w,width=2,fill=theme.value[1])
        if self.maze_map[cell]['N']==False:
            self._canvas.create_line(y, x, y + w, x,width=2,fill=theme.value[1])
        if self.maze_map[cell]['S']==False:
            self._canvas.create_line(y, x + w, y + w, x + w,width=2,fill=theme.value[1])

    _tracePathList=[]
    def _tracePathSingle(self,a,p,kill,showMarked,delay):#tracePath method for tracing a path by agent.
        def killAgent(a): #if the agent should be killed after it reaches the Goal or completes the path
            for i in range(len(a._body)):
                self._canvas.delete(a._body[i])
            self._canvas.delete(a._head)
        w=self._cell_width
        if((a.x,a.y) in self.markCells and showMarked):
            w=self._cell_width
            x=a.x*w-w+self._LabWidth
            y=a.y*w-w+self._LabWidth
            self._canvas.create_oval(y + w/2.5+w/20, x + w/2.5+w/20,y + w/2.5 +w/4-w/20, x + w/2.5 +w/4-w/20,fill='red',outline='red',tag='ov')
            self._canvas.tag_raise('ov')

        if (a.x,a.y)==(a.goal):
            del maze._tracePathList[0][0][a]
            if maze._tracePathList[0][0]=={}:
                del maze._tracePathList[0]
                if len(maze._tracePathList)>0:
                    self.tracePath(maze._tracePathList[0][0],kill=maze._tracePathList[0][1],delay=maze._tracePathList[0][2])
            if kill:
                self._win.after(300, killAgent,a)
            return
        # If path is provided as Dictionary
        if(type(p)==dict):
            if(len(p)==0):
                del maze._tracePathList[0][0][a]
                return
            a.x,a.y=p[(a.x,a.y)]
        # If path is provided as String
        if (type(p)==str):
            if(len(p)==0):
                del maze._tracePathList[0][0][a]
                if maze._tracePathList[0][0]=={}:
                    del maze._tracePathList[0]
                    if len(maze._tracePathList)>0:
                        self.tracePath(maze._tracePathList[0][0],kill=maze._tracePathList[0][1],delay=maze._tracePathList[0][2])
                if kill:
                    self._win.after(300, killAgent,a)
                return
            # recompute the agent coordinates as it moves
            move=p[0]
            if move=='E':
                if a.y+1<=self.cols: a.y+=1
            elif move=='W':
                if a.y-1>0: a.y-=1
            elif move=='N':
                if a.x-1>0: a.x-=1
            elif move=='S':
                if a.x+1<=self.rows: a.x+=1
            p=p[1:] # point the agent at the next position
        # If path is provided as List
        if (type(p)==list):
            if(len(p)==0):
                del maze._tracePathList[0][0][a]
                if maze._tracePathList[0][0]=={}:
                    del maze._tracePathList[0]
                    if len(maze._tracePathList)>0:
                        self.tracePath(maze._tracePathList[0][0],kill=maze._tracePathList[0][1],delay=maze._tracePathList[0][2])
                if kill:
                    self._win.after(300, killAgent,a)
                return
            a.x,a.y=p[0]
            del p[0]
        self._win.after(delay, self._tracePathSingle,a,p,kill,showMarked,delay)

    def tracePath(self,d,kill=False,delay=200,showMarked=False):
        #A method to trace path by agent you can provide more than one agent/path details
        self._tracePathList.append((d,kill,delay))
        if maze._tracePathList[0][0]==d:
            for a,p in d.items():
                if a.goal!=(a.x,a.y) and len(p)!=0:
                    self._tracePathSingle(a,p,kill,showMarked,delay)
    def run(self):
        self._win.mainloop() # the tkinter call that runs the app
```

## III. THE A_Star ALGORITHM AND A COMPARISON OF THE HEURISTICS

### 1. Algorithm analysis

1. The Euclidean and Manhattan heuristic functions:
- h_euclide(cell1, cell2): computes the Euclidean distance between two cells of the maze.

  + cell1, cell2: the coordinates of the two cells.
  + Returns the Euclidean value between them, used when moving in any direction is allowed.
- h_mahattan(cell1, cell2): computes the Manhattan distance between two cells of the maze.

  + cell1, cell2: the coordinates of the two cells.
  + Returns the Manhattan value, used when only the 4 directions up, down, left and right are allowed.
- h_diagonal(cell1, cells): computes the Diagonal distance between two cells of the maze.

  + cell1, cell2: the coordinates of the two cells.
  + Returns the Diagonal value, used when all 8 directions are allowed: up, down, left, right and the 4 diagonals.
2. The A* algorithm - the function aStar(m, h, start=None):
- Setting up:

 + open: a priority queue that holds the cells to check, ordered by increasing cost.
 + aPath: a dictionary that stores the path found, from the goal cell back to the start cell.
 + g_score: a dictionary that stores the cost from the start cell to the current cell.
 + f_score: a dictionary that stores the total cost from the start cell to the goal through the current cell.
- Visiting the cells:

 + While the open queue is not empty:
 + Take the current cell (currCell) with the lowest priority cost.
 + If currCell is the goal cell, leave the loop.
 + Loop over the directions (d) and check which child cells can be reached.
If the new cost is smaller than the current cost of the child cell, update the values and push the child cell into the queue.
- Tracing the path:

 + Use aPath to trace the path from the goal cell back to the start cell.
 + The result also returns the list searchPath, which records the whole search.
3. Comparing the heuristics:
The code runs the A* algorithm with different heuristic functions to compare how well they do:
Every heuristic function is used to measure the cost and the performance of the A* algorithm when finding a path in the maze.
- Conclusion:
 + The A* algorithm uses a priority queue to make the search for a path more efficient.
 + The heuristic function estimates the cost left from the current cell to the goal cell, which helps the algorithm search in the best direction.
 + Comparing how well the different heuristic functions do shows how they affect the speed and the cost of the algorithm on the maze pathfinding problem.

```python
def h_euclide(cell1, cell2): #heuristic between the coordinates of two cells
    x1, y1 = cell1
    x2, y2 = cell2
    return sqrt((x1-x2)**2+(y1-y2)**2) # Euclide
def h_mahattan(cell1, cell2): #heuristic between the coordinates of two cells
    x1, y1 = cell1
    x2, y2 = cell2
    return (abs(x1 - x2) + abs(y1 - y2)) # mahattan
def h_diagonal(cell1, cell2): #heuristic between the coordinates of two cells
    x1, y1 = cell1
    x2, y2 = cell2

    dx = abs(x1 - x2)
    dy = abs(y1 - y2)

    # dx = abs(current_cell.x - goal.x)
    # dy = abs(current_cell.y - goal.y)

    h = 1*(dx+dy) + (sqrt(2)-2*1)*min(dx,dy)
    # h = D * (dx + dy) + (D2 - 2 * D) * min(dx, dy)
    # where D is length of each node(usually = 1) and D2 is diagonal distance between each node (usually = sqrt(2) ).
    return h

def aStar(m,h,start=None): #m is a maze with rows and column
    if start is None:
        start=(m.rows,m.cols) # the start is the last cell of the maze, (1,1) is the goal
    open = PriorityQueue()
    open.put((h(start, m._goal), h(start, m._goal), start))
    aPath = {} # a dictionary of key: value <=> child : parent
    g_score = {row: float("inf") for row in m.grid} # g is the cost from the start node to the current node n
    g_score[start] = 0
    f_score = {row: float("inf") for row in m.grid}
    f_score[start] = h(start, m._goal) # for the start node, f starts as the heuristic value from the start node
                       # to the goal node; in this case the goal node is (1,1)
    searchPath=[start]
    while not open.empty(): # while the queue is not empty
        currCell=open.get()[2] # take the cell with the smallest f from the priority queue
        searchPath.append(currCell)
        if currCell == m._goal: # if that cell is the goal (1,1), break
            break
        for d in 'ESNW': # east, south, north, west
            if m.maze_map[currCell][d]==True:
                if d=='E':
                    childCell=(currCell[0],currCell[1]+1)
                elif d=='W':
                    childCell=(currCell[0],currCell[1]-1)
                elif d=='N':
                    childCell=(currCell[0]-1,currCell[1])
                elif d=='S':
                    childCell=(currCell[0]+1,currCell[1])

                temp_g_score = g_score[currCell] + 1
                temp_f_score = temp_g_score + h(childCell, m._goal) # the new f is the new g + the new h

                if temp_f_score < f_score[childCell]: #if the new f is smaller than the old f, update f and g
                    aPath[childCell] = currCell
                    g_score[childCell] = temp_g_score
                    f_score[childCell] = temp_g_score + h(childCell, m._goal)
                    open.put((f_score[childCell], h(childCell, m._goal), childCell)) # childCell is pushed into the queue
    fwdPath={} # used to reverse the order of aPath
    cell=m._goal # aPath runs goal -> start, fwdPath runs start -> goal
    while cell!=start:
        fwdPath[aPath[cell]]=cell # fwdPath(child: parent)
        cell=aPath[cell] # cell = its parent node
    return searchPath,aPath,fwdPath
```

```python
myMaze=maze(12,10)
myMaze.CreateMaze() # create the maze
searchPath,aPath,fwdPath=aStar(myMaze,h_mahattan) #h_mahattan
searchPath2,aPath2,fwdPath2=aStar(myMaze,h_euclide) #h_euclide
searchPath3,aPath3,fwdPath3=aStar(myMaze,h_diagonal) #h_diagonal


l=textLabel(myMaze,'ManhatPath',len(fwdPath)+1)
l=textLabel(myMaze,'EuclidPath',len(fwdPath2)+1)
l=textLabel(myMaze,'DiagonalPath',len(fwdPath3)+1)

l=textLabel(myMaze,'ManhatSearch',len(searchPath)+1)
l=textLabel(myMaze,'EuclidSeach',len(searchPath2)+1)
l=textLabel(myMaze,'DiagonalSeach',len(searchPath3)+1)


a=agent(myMaze,footprints=True,color=COLOR.cyan,filled=True)
b=agent(myMaze,footprints=True,color=COLOR.yellow)
c=agent(myMaze,footprints=True,color=COLOR.red)

myMaze.tracePath({a:fwdPath},delay=100)
myMaze.tracePath({b:fwdPath2},delay=100)
myMaze.tracePath({c:fwdPath3},delay=100)


t1=timeit(stmt='aStar(myMaze,h_mahattan)',number=100,globals=globals())
t2=timeit(stmt='aStar(myMaze,h_euclide)',number=100,globals=globals())
t3=timeit(stmt='aStar(myMaze,h_diagonal)',number=100,globals=globals())


textLabel(myMaze,'Manhattan Time',t1) # show a label named Manhattan Time with time t1
textLabel(myMaze,'Euclidean Time',t2) # show a label named Euclidean Time with time t2
textLabel(myMaze,'Diagonal Time',t3) # show a label named Diagonal Time with time t3


myMaze.run()
```

Using the BFS algorithm to solve the maze pathfinding problem.

### 1. Algorithm analysis

- Setting up:
    + frontier: a deque that holds the cells BFS is exploring; the start cell is the first one put into the deque
    + bfsPath: a dictionary that stores the child-parent links in BFS
    + explored : a list that tracks the cells already explored
    + bSearch=[] : a list that stores the cells BFS has explored
- Visiting the cells:
   + While the frontier is not empty, take the cell on the left of the deque
   + If that cell is not the goal yet, look at the East, West, South and North directions and add those cells to the deque
   + Repeat until the goal is reached
- Tracing:
   + bfsPath gives the path from the goal back to the start
   + Use fwdPath to store the path from the start to the goal


```python
from collections import deque
def BFS(m,start=None):  #define the BFS function
    if start is None:
        start=(m.rows,m.cols)
    #declare the variables
    frontier = deque()  #a deque holding the cells BFS is exploring
    frontier.append(start)
    bfsPath = {}  #a dictionary holding the child-parent links in BFS
    explored = [start] #a list tracking the cells already explored
    bSearch=[] #a list storing the cells BFS explored

    #BFS loops until the frontier is empty or the goal is reached
    while len(frontier)>0:
        currCell=frontier.popleft()
        if currCell==m._goal: #if the goal is reached, stop
            break
        for d in 'ESNW':
            if m.maze_map[currCell][d]==True: #for each cell in the frontier, check the neighbour cells
                                              #in the 'ESNW' directions and explore the valid ones
                if d=='E':
                    childCell=(currCell[0],currCell[1]+1)
                elif d=='W':
                    childCell=(currCell[0],currCell[1]-1)
                elif d=='S':
                    childCell=(currCell[0]+1,currCell[1])
                elif d=='N':
                    childCell=(currCell[0]-1,currCell[1])
                # if the child cell is already in the explored list, do nothing,
                if childCell in explored:
                    continue
                # otherwise add the child cell to both frontier and explored
                explored.append(childCell)
                frontier.append(childCell)
                # record the new cell as a child of the current node
                bfsPath[childCell] = currCell
                # add the new cell to the list of explored cells
                bSearch.append(childCell)
    #walk back to build the path
    #fwdPath stores the path from the goal back to the start
    fwdPath={}
    cell=m._goal #cell is set to the goal cell
    while cell!=(m.rows,m.cols): #loop backwards from the goal until the start cell is reached
        fwdPath[bfsPath[cell]]=cell
        cell=bfsPath[cell] #move the current cell to its parent on the path
    return bSearch,bfsPath,fwdPath

if __name__=='__main__':

    m=maze(17,10)
    m.CreateMaze() #create a maze of size 17x10
    bSearch,bfsPath,fwdPath=BFS(m) #run the algorithm
    #create 3 agents 'a','b','c' with different settings

    a=agent(m,footprints=True,color=COLOR.yellow,filled=True)
    b=agent(m,footprints=True,color=COLOR.red,filled=False)
    c=agent(m,1,1,footprints=True,color=COLOR.cyan,filled=True,goal=(m.rows,m.cols))
    m.tracePath({a:bSearch},delay=100)#every cell BFS went through
    m.tracePath({c:bfsPath},delay=100)#the path from the goal to the start
    m.tracePath({b:fwdPath},delay=100)#the path from the start to the goal
    l=textLabel(m,'Length of Shortest Path',len(bfsPath)+1)#show the length of the shortest path on screen
    l=textLabel(m,'Length of Shortest Path',len(bSearch)+1)
    m.run()
```

## V. THE DFS ALGORITHM

<div style="text-align:center;">
    <img src="https://drive.google.com/uc?id=1rnNd3iR36uqx5l-baBjVcRRrbou-i0_W" alt="Alt text" width="600" height = 500/>
</div>

```python
def DFS(m,start=None):
    if start is None:
        start=(m.rows,m.cols)
    explored=[start] #the cells already explored
    frontier=[start] #the stack used for the DFS walk
    dfsPath={}       #the path
    dSeacrh=[]       # records the order in which the cells were explored
    #dsdsa
    while len(frontier)>0: #loop until the stack is empty
        currCell=frontier.pop()  # pop a cell off the top of the stack and record it in dSeacrh.
        dSeacrh.append(currCell)
        if currCell==m._goal: # stop when the goal is reached
            break
        poss=0 #counts how many directions DFS can go from the current cell
        for d in 'ESNW':#loop over the possible directions ('ESNW') and check whether the
                        #neighbour in that direction is valid and not yet explored.
            if m.maze_map[currCell][d]==True:
                if d =='E':
                    child=(currCell[0],currCell[1]+1)
                if d =='W':
                    child=(currCell[0],currCell[1]-1)
                if d =='N':
                    child=(currCell[0]-1,currCell[1])
                if d =='S':
                    child=(currCell[0]+1,currCell[1])
                if child in explored:
                    continue
                poss+=1
                explored.append(child)
                frontier.append(child)
                #this line makes currCell (the current cell)
                #the parent of child in dfsPath, recording the links between cells in the search tree.
                dfsPath[child]=currCell
        if poss>1:
            m.markCells.append(currCell)
    fwdPath={}
    cell=m._goal
    #while loop that reads out the path from the start to the goal
    while cell!=start:
        fwdPath[dfsPath[cell]]=cell
        cell=dfsPath[cell]

    return dSeacrh,dfsPath,fwdPath

if __name__=='__main__':
    m=maze(10,10) # Change to any size
    m.CreateMaze(2,4) # (2,4) is Goal Cell, Change that to any other valid cell

    dSeacrh,dfsPath,fwdPath=DFS(m,(5,1)) # (5,1) is Start Cell, Change that to any other valid cell
    #create 3 agents
    #a: all the states the agent can reach
    #b: shows the path from the goal to the start
    #c: shows the path from the start to the goal
    a=agent(m,5,1,goal=(2,4),footprints=True,color=COLOR.green)
    b=agent(m,2,4,goal=(5,1),footprints=True,filled=True)
    c=agent(m,5,1,footprints=True,color=COLOR.yellow)
    m.tracePath({a:dSeacrh},showMarked=True)
    m.tracePath({b:dfsPath})
    m.tracePath({c:fwdPath})
    m.run()
```

## VI. COMPARING THE BFS AND DFS ALGORITHMS
  1. Path length:

- BFS: always finds the shortest path. So even if the goal sits at the deepest level, BFS will find the shortest path

- DFS: DFS does not guarantee the shortest path. It can find a path, but not necessarily the shortest one

2. Memory space:

- BFS: has to store all the nodes of the same level in the queue. It therefore needs more memory than DFS, especially on large graphs and on mazes with many branches

- DFS: needs less memory because it only has to store one path at a time. It can run into trouble, though, when the graph or the maze has endless cycles

## VII. THE DIJKSTRA ALGORITHM


<div style="text-align:center;">
    <img src="https://drive.google.com/uc?id=1Ts7JQDz4JFXZYBrRqYsHQGLS7HRSjzF8" alt="Alt text" width="700" height = 500/>
</div>

```python
from pyamaze import maze,agent,COLOR,textLabel
def dijkstra(m,*h,start=None): #*h is a list <=> it takes any number of arguments
    if start is None:
        start=(m.rows,m.cols)

    hurdles=[(i.position,i.cost) for i in h] #build the list of hurdles from the list of agents (h),
                                             #where each hurdle holds a position and a cost.

    unvisited={n:float('inf') for n in m.grid} # start every unvisited cost at infinity
    unvisited[start]=0 # set the cost of the start node to 0.
    visited={} # the visited nodes
    revPath={} # stores the path used to walk back
    while unvisited: # loop until every node has been visited
        currCell=min(unvisited,key=unvisited.get) # pick the node with the smallest cost from
                                                  # unvisited as the current node.
        visited[currCell]=unvisited[currCell] # add the current node to the visited list
        if currCell==m._goal: # if the current node is the goal, end the loop
            break
        for d in 'EWNS':  # loop over the directions (East, West, South, North).
            if m.maze_map[currCell][d]==True: # if that direction is open
            # work out the child node from direction 'd'
                if d=='E':
                    childCell=(currCell[0],currCell[1]+1)
                elif d=='W':
                    childCell=(currCell[0],currCell[1]-1)
                elif d=='S':
                    childCell=(currCell[0]+1,currCell[1])
                elif d=='N':
                    childCell=(currCell[0]-1,currCell[1])
                if childCell in visited: ## if the child node is already visited, skip it
                    continue
                tempDist= unvisited[currCell]+1 # a temporary variable for the cost to the child node
                for hurdle in hurdles: # go through the hurdle list and add their cost
                    if hurdle[0]==currCell:
                        tempDist+=hurdle[1]
                # if the new cost to the child node is smaller, update that child node
                if tempDist < unvisited[childCell]:
                    unvisited[childCell]=tempDist
                    revPath[childCell]=currCell
        unvisited.pop(currCell) #remove the current node from the unvisited list.

    fwdPath={} #a dictionary that stores the path from the start node to the last node.
    cell=m._goal #set the goal node
    # walk the path backwards
    while cell!=start:
        fwdPath[revPath[cell]]=cell
        cell=revPath[cell]

    return fwdPath,visited[m._goal]

# run the code
if __name__=='__main__':
    # create the maze
    myMaze=maze(20,15)
    myMaze.CreateMaze(1,4)

    h1=agent(myMaze,4,4,color=COLOR.red)
    h2=agent(myMaze,4,6,color=COLOR.red)
    h3=agent(myMaze,4,1,color=COLOR.red)
    h4=agent(myMaze,4,2,color=COLOR.red)
    h5=agent(myMaze,4,3,color=COLOR.red)

    h1.cost=100
    h2.cost=100
    h3.cost=100
    h4.cost=100
    h5.cost=100

    # path,c=dijstra(myMaze,h1,h2,h2,h3,h4,h5)
    path,c=dijkstra(myMaze,h1,h2,h3,h4,h5,start=(16,1))
    textLabel(myMaze,'Total Cost',c)

    # a=agent(myMaze,color=COLOR.cyan,filled=True,footprints=True)
    a=agent(myMaze,16,1,color=COLOR.cyan,filled=True,footprints=True)
    myMaze.tracePath({a:path})

    myMaze.run()
```
